```python
import mediapipe as mp
import cv2
```

```python
mp_hands = mp.solutions.hands
mp_draw = mp.solutions.drawing_utils
```

```python
data = mp_hands.Hands(max_num_hands=1)
```

```python
video = cv2.VideoCapture(0)
while True:
    success,image = video.read()
    imag1 = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
    result = data.process(imag1)
    tipids = [4,8,12,16,20]
    lmlist = []
    if result.multi_hand_landmarks:
        for x in result.multi_hand_landmarks:
            for id,lm in enumerate(x.landmark):
                # print(id,lm)
                cx = lm.x 
                cy = lm.y
                lmlist.append([id,cx,cy])

                if len(lmlist)!=0 and len(lmlist)==21:
                    fingerlst = []

                    # Thumb 

                    if lmlist[12][1]<lmlist[20][1]:
                        if lmlist[4][1]>lmlist[3][1]:
                            fingerlst.append(0)
                        else:
                            fingerlst.append(1)

                    else:
                        if lmlist[4][1]<lmlist[3][1]:
                            fingerlst.append(0)
                        else:
                            fingerlst.append(1) 

                    # All finger except thumb

                    for i in range(1,5): 
                        if lmlist[tipids[i]][2]>lmlist[tipids[i]-2][2]:
                            fingerlst.append(0)
                        else:
                            fingerlst.append(1)
                    # print(fingerlst)
                    fingercount =fingerlst.count(1)
                    # print(fingercount)
                    cv2.putText(image,"Count: "+str(fingercount),(35,400),cv2.FONT_HERSHEY_COMPLEX,2,(0,0,255),3)

                # print(lmlist)
            mp_draw.draw_landmarks(image,x,mp_hands.HAND_CONNECTIONS,
                                   mp_draw.DrawingSpec(color=(255,0,0),thickness=2),
                                   mp_draw.DrawingSpec(color=(0,0,255),thickness=2))

    cv2.imshow("Hand",image)
    if cv2.waitKey(1)&0XFF==ord('q'):
        break

video.release()
cv2.destroyAllWindows()
```
